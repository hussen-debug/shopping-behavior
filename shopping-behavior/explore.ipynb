# FlashFash Shopping Behavior Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/shopping.csv')
df.head()
```

## Question 1: What is the effect of promo codes on the dollar amount of purchases?

```python
# group by promo code and get the average purchase amount for each group
promo_groups = df.groupby('Promo Code Used')
promo_avg = promo_groups['Purchase Amount (USD)'].mean()
print(promo_avg)

# bar chart comparing average spend
plt.bar(promo_avg.index, promo_avg.values, color=['steelblue', 'coral'], edgecolor='black')
plt.title('Average Purchase Amount by Promo Code Usage')
plt.xlabel('Promo Code Used')
plt.ylabel('Average Purchase Amount (USD)')
plt.show()
```

Customers who used a promo code spent an average of $50.02, while customers who didn't spent $30.17. That's almost a $20 difference. Promo codes are pulling in bigger purchases, not just discounting small ones.

## Question 2: What are the most popular colors by season?

```python
# count how many times each color appears per season
color_counts = df.groupby(['Season', 'Color']).size().reset_index(name='Count')

# sort and grab top 3 per season
color_counts = color_counts.sort_values(['Season', 'Count'], ascending=[True, False])
top_colors = color_counts.groupby('Season').head(3)

print(top_colors.to_string(index=False))
```

The color trends make a lot of sense. Fall goes heavy on earth tones like Brown and Burnt Orange. Winter leans dark with Black and Aubergine. Summer is all pastels like Lavender and Lemon Yellow, and Spring sticks to lighter colors like Baby Blue and Pale Peach. Inventory planning can follow these patterns directly.

## Question 3: What is the most popular clothing item by season?

```python
# count purchases per item per season
item_counts = df.groupby(['Season', 'Item Purchased']).size().reset_index(name='Count')

# sort and get the top item per season
item_counts = item_counts.sort_values(['Season', 'Count'], ascending=[True, False])
top_items = item_counts.groupby('Season').first().reset_index()

print(top_items[['Season', 'Item Purchased', 'Count']].to_string(index=False))

# bar chart with item labels above each bar
seasons = top_items['Season'].tolist()
counts = top_items['Count'].tolist()
items = top_items['Item Purchased'].tolist()

plt.bar(seasons, counts, color=['#c8a165', '#90c97e', '#f4a261', '#6fa8d6'], edgecolor='black')

for i in range(len(seasons)):
    plt.text(i, counts[i] + 1, items[i], ha='center', fontsize=9)

plt.title('Most Popular Item Per Season')
plt.xlabel('Season')
plt.ylabel('Number of Purchases')
plt.show()
```

Every season has a clear winner. Socks in Fall, Running Shoes in Spring, Shorts in Summer, and Leggings in Winter. Summer Shorts came in highest overall with 177 purchases. Stocking up on these items before each season starts is a straightforward win.

## Question 4: Which age groups spend the most, and does that change by season?

```python
# create age buckets
df['Age Group'] = pd.cut(df['Age'], bins=[17, 25, 35, 50, 75], labels=['18-25', '26-35', '36-50', '51+'])

# average purchase amount per age group and season
age_season = df.groupby(['Age Group', 'Season'])['Purchase Amount (USD)'].mean().unstack().round(2)
print(age_season)

# grouped bar chart
age_season.plot(kind='bar', edgecolor='black')
plt.title('Average Purchase Amount by Age Group and Season')
plt.xlabel('Age Group')
plt.ylabel('Average Purchase Amount (USD)')
plt.xticks(rotation=0)
plt.legend(title='Season')
plt.show()
```

Spending is pretty consistent across all age groups, usually falling between $37 and $43. The one standout is the 51+ group in Spring who average $42.57, which is the highest across the whole table. Age doesn't seem to be a major driver of spending in this dataset, but older shoppers in Spring might be worth targeting.

## Question 5: When do users leave a review?

```python
# flag rows that have a review rating
df['Has Review'] = df['Review Rating'].notna()

# filter out rows with no frequency data
has_freq = df[df['Frequency of Purchases'].notna()]

# get review rate per frequency group
review_rate = has_freq.groupby('Frequency of Purchases')['Has Review'].mean()
review_pct = (review_rate * 100).round(1)
review_pct = review_pct.sort_values(ascending=False)

print(review_pct)

# bar chart
plt.bar(review_pct.index, review_pct.values, color='mediumpurple', edgecolor='black')
plt.title('Review Rate by Purchase Frequency')
plt.xlabel('Frequency of Purchases')
plt.ylabel('Percent Who Left a Review')
plt.xticks(rotation=30)
plt.show()
```

Weekly shoppers leave reviews the most often at 48.3%. Bi-Weekly shoppers are the least likely at 35.5%. No group breaks 50% though, so most purchases go unreviewed. Some kind of follow-up prompt after a purchase could help bring those numbers up.
